Derivation of following equation with respect to `w`

$$
\frac{1}{2}
\mathbf{w}^T\mathbf{\Phi}^T\mathbf{\Phi}\mathbf{w}
$$


**Step 1: Recall the general matrix differentiation rule.**

For a square matrix $A$, we have:

$$
\frac{\partial}{\partial \mathbf{w}}
\left(
\frac{1}{2}\mathbf{w}^T A\mathbf{w}
\right)
=
\frac{1}{2}(A+A^T)\mathbf{w}
$$

**Step 2: Identify the matrix $A$.**

Comparing the given expression with the general form, we identify:

$$
A=\mathbf{\Phi}^T\mathbf{\Phi}
$$

Taking its transpose:

$$
\begin{aligned}
A^T
&=(\mathbf{\Phi}^T\mathbf{\Phi})^T\\
&=\mathbf{\Phi}^T(\mathbf{\Phi}^T)^T\\
&=\mathbf{\Phi}^T\mathbf{\Phi}\\
&=A
\end{aligned}
$$

Thus, $A$ is a symmetric matrix.

**Step 3: Apply the differentiation rule.**

Substituting $A=\mathbf{\Phi}^T\mathbf{\Phi}$ into the general rule:

$$
\begin{aligned}
&\frac{\partial}{\partial \mathbf{w}}
\left(
\frac{1}{2}\mathbf{w}^T
\mathbf{\Phi}^T\mathbf{\Phi}
\mathbf{w}
\right)\\
&=\frac{1}{2}
\left(
\mathbf{\Phi}^T\mathbf{\Phi}
+
(\mathbf{\Phi}^T\mathbf{\Phi})^T
\right)\mathbf{w}\\
&=\frac{1}{2}
\left(
\mathbf{\Phi}^T\mathbf{\Phi}
+
\mathbf{\Phi}^T\mathbf{\Phi}
\right)\mathbf{w}\\
&=\frac{1}{2}
\left(
2\mathbf{\Phi}^T\mathbf{\Phi}
\right)\mathbf{w}
\end{aligned}
$$

Therefore, the final result is:

$$
\boxed{
\frac{\partial}{\partial \mathbf{w}}
\left(
\frac{1}{2}\mathbf{w}^T
\mathbf{\Phi}^T\mathbf{\Phi}
\mathbf{w}
\right)
=
\mathbf{\Phi}^T\mathbf{\Phi}\mathbf{w}
}
$$
