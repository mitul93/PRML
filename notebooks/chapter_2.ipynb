## Binary Variables

Consider a single binary random variable $x \in {0,1}$

$$
\begin{align}
p(x=1 | \mu) &= \mu \newline
Bern(x|\mu) &= \mu^x (1 - \mu)^{1-x}
\end{align}
$$

$$
\begin{align}
E[x] &= \mu \newline
var[x] &= \mu (1-\mu)
\end{align}
$$

Assume we have data $D = {x_1, x_2, ..., x_n}$ of observed values of $x$.
Likelihood function, which is function of $\mu$ assuming observations are drawn independently from $p(x|\mu)$, so that

$$
p(D|\mu) = \sum_{n=1}^{N} p(x_n | \mu) = \sum_{n=1}^N \mu^x_n (1 - \mu)^{1-x_n}
$$

Log Likelihood,

$$
\ln{p(D|\mu)} = \sum_{n=1}^{N} \ln{p(x_n | \mu)} = \sum_{n=1}^{N} {x_n \ln{\mu} + (1-x_n) \ln({1-\mu})}
$$
